```python
def q(df, cat, num, aggre):

    # Calculate IQR
    q1 = df[num].quantile(0.25)
    q3 = df[num].quantile(0.75)
    iqr = q3 - q1

    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    # Histplot
    plt.figure(figsize=(12, 9))
    sns.histplot(df[num], kde=True)
    plt.title(f'Histplot Of {num} Lower Bound: {lower} Upper Bound: {upper}')
    plt.show()

    # Box Plot
    plt.figure(figsize=(12, 9))
    sns.boxplot(x=df[num])
    plt.title(f'Box Plot Of {num} Lower Bound: {lower} Upper Bound: {upper}')
    plt.show()

    # Groupby
    z = df.groupby(cat, as_index=False)[num].agg(aggre).sort_values(
        by=num, ascending=False
    )

    # Pie Chart
    if df[cat].nunique() <= 4:
        plt.figure(figsize=(8, 8))
        plt.pie(
            z[num],
            labels=z[cat],
            autopct='%1.1f%%'
        )
        plt.title(f'{aggre} {num} by {cat}')

    # Line Chart
    elif df[cat].nunique() in [7, 12, 24, 30, 31]:
        plt.figure(figsize=(12, 6))
        sns.lineplot(
            data=z,
            x=cat,
            y=num,
            marker='o'
        )
        plt.grid(True)
        plt.title(f'{aggre} {num} by {cat}')

    # Bar Chart
    elif df[cat].nunique() >= 10:
        plt.figure(figsize=(12, 6))
        sns.barplot(
            data=z,
            x=cat,
            y=num
        )
        plt.title(f'{aggre} {num} by {cat}')

    # Horizontal Bar Chart
    elif df[cat].nunique() <= 8:
        plt.figure(figsize=(12, 6))
        sns.barplot(
            data=z,
            x=num,
            y=cat
        )
        plt.title(f'{aggre} {num} by {cat}')

    plt.show()
```

```python
def outlier(df,num): #automation using charts for outliers
    # Histplot
    q1=df[num].quantile(0.25)
    q3=df[num].quantile(0.75)
    avg=round(df[num].mean(),2)
    median=df[num].median()
    mode=df[num].mode()
    Min=df[num].min()
    Max=df[num].max()
    iqr=q3-q1

    
    plt.figure(figsize=(12,9))
    sns.histplot(df[num], kde=1)
    plt.title(f'Distrobution Of {num} \n Lower Bound:{q1-1.5*iqr} \n Upper Bound:{q3+1.5*iqr} \n Average:{avg} \n Median:{median} \n Mode:{mode} \n Q1:{q1} \n Q3:{q3} \n Min:{Min} \n Max:{Max}')
    plt.grid()
    plt.show()

    # Box Plot
    plt.figure(figsize=(12,9))
    sns.boxplot(df[num])
    plt.title(f'Box Plot Of {num} \n Lower Bound:{q1-1.5*iqr} \n Upper Bound:{q3+1.5*iqr} \n Average:{avg} \n Median:{median} \n Mode:{mode} \n Q1:{q1} \n Q3:{q3} \n Min:{Min} \n Max:{Max}')
    plt.grid()
    plt.show()
    return 
```

```python
plt.figure(figsize=(12, 9))

ax = sns.boxplot(y=df['Patient  Age'])

# -------------------------
# Calculate values
# -------------------------
q1 = df['Patient  Age'].quantile(0.25)
q3 = df['Patient  Age'].quantile(0.75)
median = df['Patient  Age'].median()
mean = df['Patient  Age'].mean()
mode = df['Patient  Age'].mode()[0]
min_val = df['Patient  Age'].min()
max_val = df['Patient  Age'].max()

IQR = q3 - q1
lower_bound = q1 - 1.5 * IQR
upper_bound = q3 + 1.5 * IQR


# -------------------------
# Labels with arrows
# -------------------------

labels = [
    ('Min', min_val, 0.35, min_val - 2),
    ('Q1', q1, 0.35, q1 - 2),
    ('Median', median, 0.35, median + 3),
    ('Mean', round(mean, 2), -0.65, mean + 5),
    ('Mode', mode, 0.35, mode + 2),
    ('Q3', q3, 0.35, q3 + 2),
    ('Max', max_val, 0.35, max_val + 1)
]

for label, value, x, y in labels:
    ax.annotate(
        f'{label}: {value}',
        xy=(0, value),
        xytext=(x, y),
        arrowprops={'arrowstyle': '->'},
        fontsize=10
    )



plt.text(
    0.98, 0.02,
    f'Lower Bound: {lower_bound:.1f}\n'
    f'Upper Bound: {upper_bound:.1f}',
    transform=ax.transAxes,
    ha='right',
    va='bottom',
    fontsize=10
)

plt.title('Box Plot of Patient Age')
plt.ylabel('Patient  Age')

plt.show()
```
